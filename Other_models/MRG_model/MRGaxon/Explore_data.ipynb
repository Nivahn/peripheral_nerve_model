```python
import pandas as pd
import matplotlib.pyplot as plt

# Read the CSV file without headers
csv_50Hz = "./../../../Data/SCS_LTMRs_50Hz_Timestamps.csv"
try:
    # Read CSV without headers - each column represents a different neuron
    df = pd.read_csv(csv_50Hz, header=None)

    # Print basic info about the data
    print(f"Number of neurons: {len(df.columns)}")
    print(f"Shape of data: {df.shape}")
    print(f"First few rows:\n{df.head()}")

except FileNotFoundError:
    print(f"File not found at: {csv_50Hz}")
    exit()
except Exception as e:
    print(f"Error reading file: {e}")
    exit()

# Create the raster plot
plt.figure(figsize=(12, 8))

# Plot each neuron's spikes
for neuron_idx in range(len(df.columns)):
    # Get spike times for this neuron (remove NaN values)
    spike_times = df[neuron_idx].dropna().values

    # Create vertical lines for each spike
    plt.eventplot(spike_times, lineoffsets=neuron_idx+1, linelengths=0.8,
                  color='blue', linewidths=0.4)

# Customize the plot
plt.xlim(0, 5)
plt.xlabel('Time', fontsize=12)
plt.ylabel('', fontsize=12)
plt.title('LTMRs during 50-Hz SCS', fontsize=14)
plt.yticks(range(1, len(df.columns) + 1), [f'Neuron {i+1}' for i in range(len(df.columns))])
plt.grid(True, alpha=0.3, linestyle='--')

# Adjust layout and show
plt.tight_layout()
plt.show()

# Print spike statistics
print("\nSpike statistics:")
for neuron_idx in range(len(df.columns)):
    spike_count = df[neuron_idx].count()
    if spike_count > 0:
        spike_times = df[neuron_idx].dropna()
        print(f"Neuron {neuron_idx+1}: {spike_count} spikes, "
              f"Time range: {spike_times.min():.2f} - {spike_times.max():.2f}")
    else:
        print(f"Neuron {neuron_idx+1}: No spikes")
```

```python
import h5py
import numpy as np
from scipy.signal import find_peaks

Data_file_50Hz = "./../../../Data/MRG_responses_50Hz.h5"

after_branch_daughter = []
after_branch_main = []
before_branch = []
stimulation_point = []

with h5py.File(Data_file_50Hz, 'r') as f:
    print("Top-level keys:", list(f.keys()))

    # Возьмём время из первой группы нейрона – оно одинаковое для всех
    example_neuron = [k for k in f.keys() if k.startswith("Neuron_")][0]
    time = f[f"{example_neuron}/Model/time"][:]  # в мс (как ты сам записывал)
    dt = time[1] - time[0]
    print(f"dt (model) ≈ {dt} ms, time range: {time[0]}–{time[-1]} ms")

    for neuron in f.keys():
        if not neuron.startswith("Neuron_"):
            continue

        print(f"\nNeuron group: {neuron}")
        traces_group = f[f"{neuron}/Model/Traces"]

        # ---- before_branch ----
        if "before_branch" in traces_group:
            grp = traces_group["before_branch"]
            # возьмём первый node (если их несколько)
            first_node_name = sorted(grp.keys())[0]
            signal = grp[first_node_name][:]

            peaks, _ = find_peaks(signal, distance=int(1.0/dt))
            spike_times_ms = time[peaks]
            before_branch.append(spike_times_ms)
            print(f"  before_branch: {len(spike_times_ms)} spikes")

        # ---- after_branch_main ----
        if "after_branch_main" in traces_group:
            grp = traces_group["after_branch_main"]
            first_node_name = sorted(grp.keys())[0]
            signal = grp[first_node_name][:]

            peaks, _ = find_peaks(signal, distance=int(1.0/dt))
            spike_times_ms = time[peaks]
            after_branch_main.append(spike_times_ms)
            print(f"  after_branch_main: {len(spike_times_ms)} spikes")

        # ---- after_branch_daughter ----
        if "after_branch_daughter" in traces_group:
            grp = traces_group["after_branch_daughter"]
            first_node_name = sorted(grp.keys())[0]
            signal = grp[first_node_name][:]

            peaks, _ = find_peaks(signal, distance=int(1.0/dt))
            spike_times_ms = time[peaks]
            after_branch_daughter.append(spike_times_ms)
            print(f"  after_branch_daughter: {len(spike_times_ms)} spikes")

        # ---- stimulation_point ---- (если есть)
        if "stimulation_point" in traces_group:
            grp = traces_group["stimulation_point"]
            first_node_name = sorted(grp.keys())[0]
            signal = grp[first_node_name][:]

            peaks, _ = find_peaks(signal, distance=int(1.0/dt))
            spike_times_ms = time[peaks]
            stimulation_point.append(spike_times_ms)
            print(f"  stimulation_point: {len(spike_times_ms)} spikes")
```

```python
import pandas as pd

csv_50Hz = "./../../../Data/SCS_LTMRs_50Hz_Timestamps.csv"

df = pd.read_csv(csv_50Hz, header=None)
print(f"Number of neurons (input): {df.shape[1]}")
print(f"First rows:\n{df.head()}")

input_spikes_ms = []
for neuron_idx in range(df.shape[1]):
    spike_times_s = df[neuron_idx].dropna().values  # предположим, здесь секунды
    spike_times_ms = spike_times_s * 1000.0
    input_spikes_ms.append(spike_times_ms)
```

```python
import matplotlib.pyplot as plt

n_inputs = len(input_spikes_ms)
n_model = len(before_branch)  # должно совпадать с числом нейронов

fig, axes = plt.subplots(1, 4, figsize=(24, 10), sharey=True)

# ---------- Панель 1: входные нейроны ----------
ax = axes[0]
ax.eventplot(
    input_spikes_ms,
    lineoffsets=np.arange(1, n_inputs + 1),
    linelengths=0.8,
    colors='blue',
    linewidths=0.4
)
ax.set_title("Вход: LTMR (50 Hz SCS)")
ax.set_xlabel("Время (мс)")
ax.set_ylabel("Нейрон")
ax.set_xlim(0, 5000)  # первые 5 секунд
ax.set_yticks(range(1, n_inputs + 1))
ax.set_yticklabels([f"N{idx+1}" for idx in range(n_inputs)])
ax.grid(True, alpha=0.3, linestyle='--')

# ---------- Панель 2: до ветвления ----------
ax = axes[1]
ax.eventplot(
    before_branch,
    lineoffsets=np.arange(1, n_model + 1),
    linelengths=0.8,
    colors='green',
    linewidths=0.4
)
ax.set_title("До ветвления (main axon)")
ax.set_xlabel("Время (мс)")
ax.set_xlim(0, 5000)
ax.grid(True, alpha=0.3, linestyle='--')

# ---------- Панель 3: после ветвления — основная ветвь ----------
ax = axes[2]
ax.eventplot(
    after_branch_main,
    lineoffsets=np.arange(1, n_model + 1),
    linelengths=0.8,
    colors='red',
    linewidths=0.4
)
ax.set_title("После ветвления — основная ветвь")
ax.set_xlabel("Время (мс)")
ax.set_xlim(0, 5000)
ax.grid(True, alpha=0.3, linestyle='--')

# ---------- Панель 4: после ветвления — дочерняя ветвь ----------
ax = axes[3]
ax.eventplot(
    after_branch_daughter,
    lineoffsets=np.arange(1, n_model + 1),
    linelengths=0.8,
    colors='purple',
    linewidths=0.4
)
ax.set_title("После ветвления — дочерняя ветвь")
ax.set_xlabel("Время (мс)")
ax.set_xlim(0, 5000)
ax.grid(True, alpha=0.3, linestyle='--')

plt.suptitle("Проведение спайков от LTMR → MRG-аксон с ветвлением", fontsize=18, y=0.98)
plt.tight_layout()
plt.show()
```
