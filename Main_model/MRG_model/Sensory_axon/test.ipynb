```python



axon = MRGaxon(
    fiber_diameter=5.7,
    parent_axon_nodes=42,
    branch_nodes=10,
    branches_num=5,
    nodes_dist=5,
    diam_scale=0.6,
    celsius=37.0,
    dt_ms=dt,
    v_init=v_init,
    h_stop=h_stop_ms
)
```

```python
axon.set_stimulation_params(mode='create',
                            freq_hz=10,
                            amp=amp_nA,
                            t_start=10,
                            t_end=h_stop_ms,
                            phase_us=40.0,
                            gap_us=5,
                            plot_duration=1)
```

```python
axon.run_simulation()
```
